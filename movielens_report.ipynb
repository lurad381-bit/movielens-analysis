# Анализ данных MovieLens


## Цель работы

Цель данного задания - научиться работать с табличными данными без использования библиотеки pandas, используя только базовые средства Python.

Для анализа используются несколько файлов датасета, каждый из которых содержит информацию определенного типа:

- **movies.csv** - сведения о фильмах (идентификатор фильма, название и жанры);
- **ratings.csv** - оценки, выставленные пользователями фильмам;
- **tags.csv** - пользовательские теги, характеризующие фильмы;
- **links.csv** - соответствия идентификаторов MovieLens с идентификаторами IMDb и TMDb.

Для файла **movies.csv** используются первые 1000 записей, тогда как остальные файлы анализируются полностью.

# Структура проекта

Для работы с датасетом были предусмотрены следующие классы:

### Movies

Класс предназначен для анализа файла **movies.csv**. Позволяет исследовать распределение фильмов по годам выпуска и жанрам, а также получать список фильмов с наибольшим количеством жанров.

Класс предназначен для анализа файла **ratings.csv**. Используется для обработки пользовательских оценок фильмов, вычисления статистики рейтингов и определения наиболее популярных фильмов.

Класс предназначен для анализа файла **tags.csv**. Позволяет исследовать пользовательские теги: определять самые популярные, самые длинные, содержащие наибольшее количество слов, а также выполнять поиск тегов по заданному слову.

### Links

Класс работает с файлом **links.csv** и содержит информацию о соответствии идентификаторов MovieLens, IMDb и TMDb. Этот класс позволяет связывать данные между различными источниками.

### PyTest

Для проверки корректности работы реализованных методов были написаны автоматические тесты с использованием библиотеки **PyTest**.

# Анализ данных о связях фильмов (Links)

Каждая строчка в `links.csv` - это всего три числа: `movieId`, `imdbId`, `tmdbId`.
На первый взгляд скучная таблица идентификаторов. Но за каждым `imdbId` стоит
настоящая страница IMDB - со бюджетом, кассой, режиссёром и хронометражем.

В этом разделе мы используем класс `Links`, чтобы превратить безликие номера
в истории: кто снял больше всего фильмов из золотой эпохи 90-х? Какой фильм
обернулся самым громким финансовым провалом? И сколько стоила одна минута
экранного времени в самом дорогом фильме нашей выборки?

Класс умеет получать эти данные двумя способами: либо скрейпя страницы IMDB
напрямую (что упирается в анти-бот защиту сайта), либо читая заранее собранный
файл `imdb_cache.json` - наш локальный кэш с данными по 1000 фильмам. Именно
этот кэш и используется ниже, чтобы отчёт можно было пересчитать в любой
момент без сети.

## Загрузка данных

Прежде чем задавать вопросы, нужно создать объект `Links`. В конструкторе
происходит главное: данные из `links.csv` и `movies.csv` объединяются сразу,
и каждая запись в `links.data` получает не только `movieId`/`imdbId`/`tmdbId`,
но и настоящее название фильма.

```python
%timeit -n 1 -r 1 links = movielens_analysis.Links("links.csv")
links = movielens_analysis.Links("links.csv")
print(f"Загружено фильмов: {movies.get_count()}")
print("Пример объединённой записи:", links.data[0])
```

## Что вообще можно узнать про один фильм?

Прежде чем считать рейтинги и суммы, посмотрим на сам строительный блок -
метод `get_imdb()`. Он принимает список `movieId` и список нужных полей, а
возвращает таблицу, отсортированную по `movieId` по убыванию.

Спросим про три первых фильма в датасете: кто их снял, во что они обошлись, сколько собрали в прокате и их длительность

```python
%timeit -n 1 -r 1 links.get_imdb([1, 2, 3], ["Director", "Budget", "Cumulative Worldwide Gross", "Runtime"])
links.get_imdb([1, 2, 3], ["Director", "Budget", "Cumulative Worldwide Gross", "Runtime"])
```

Обратите внимание на порядок: `movieId=3` идёт первым, `movieId=1` -
последним. Это не баг, а спецификация метода - результат всегда сортируется
по `movieId` по убыванию.

В выборке 1000 самых первых фильмов MovieLens было бы интересно узнать,

Количество фильмов - это ещё не всё. Бонусный метод `average_budget_by_director()`

# Анализ пользовательских оценок (Ratings)

Если `Links` рассказывал про деньги и режиссёров, то `Ratings` - про самих
зрителей. Класс `Ratings` объединяет `ratings.csv` (кто, что и насколько
высоко оценил) с `movies.csv` (что это вообще был за фильм), и предоставляет
два внутренних "взгляда" на эти данные: `Ratings.Movies` - статистика по
фильмам, и `Ratings.Users` - статистика по пользователям (`Users` унаследован
от `Movies`, просто группирует те же данные по другому ключу).

Используем эти данные, чтобы выяснить: какие фильмы любят дольше всего,
какие жанры оценивают строже, и есть ли среди пользователей такие, кто ставит
только пятёрки.

## Загрузка и объединение данных

Конструктор `Ratings` сразу связывает каждую оценку с названием и жанрами
фильма - обращаться к `movies.csv` отдельно для каждого вопроса не придётся.

```python
%timeit -n 1 -r 1 ratings = movielens_analysis.Ratings("ratings.csv", "movies.csv")
ratings = movielens_analysis.Ratings("ratings.csv", "movies.csv")
movies_ratings = movielens_analysis.Ratings.Movies(ratings)
users_ratings = movielens_analysis.Ratings.Users(ratings)
print(f"Объединённых записей (оценка + фильм): {ratings.get_count()}")
```

## Как распределены оценки в целом?


```python
%timeit -n 1 -r 1 movies_ratings.dist_by_rating()
movies_ratings.dist_by_rating()
```

Оценка 4.0 - самая популярная, а следом за ней 3.0 и 5.0.

## В каком году было выставлено больше всего оценок?


```python
%timeit -n 1 -r 1 movies_ratings.dist_by_year()
movies_ratings.dist_by_year()
```

1996 год - безоговорочный пик активности, более 6000 оценок за один
год.

## Какие фильмы оценивали чаще всего?

Количество оценок - это, по сути, метрика популярности: фильм, который
оценили 300 раз, явно посмотрело гораздо больше людей, чем тот, что оценили
дважды.

```python
%timeit -n 1 -r 1 movies_ratings.top_by_num_of_ratings(10)
movies_ratings.top_by_num_of_ratings(10)
```

"Форрест Гамп" возглавляет список с большим отрывом - почти 330 оценок.

## А какие фильмы оценивают выше всего?

Здесь нужно быть осторожным: метод `top_by_ratings()` смотрит на средний или
медианный рейтинг, не на количество оценок. Это значит, что фильм, который
оценил один-единственный восторженный зритель на 5.0, формально обойдёт
признанную классику с сотнями оценок. Сравним среднее и медиану.

```python
%timeit -n 1 -r 1 movies_ratings.top_by_ratings(10, metric="average")
movies_ratings.top_by_ratings(10, metric="average")
```

В топе по среднему баллу оказались малоизвестные картины с буквально одной-двумя оценками.

```python
%timeit -n 1 -r 1 movies_ratings.top_by_ratings(10, metric="median")
movies_ratings.top_by_ratings(10, metric="median")
```

## Какие фильмы вызывают больше всего споров?

Метод `top_controversial()` считает дисперсию оценок - чем она выше, тем
сильнее мнения зрителей расходятся: для одних фильм шедевр, для других
провал.

```python
%timeit -n 1 -r 1 movies_ratings.top_controversial(10)
movies_ratings.top_controversial(10)
```

## А что насчёт самих пользователей?

`Ratings.Users` - это тот же набор методов, что и у `Movies`, только
группировка идёт по `userId`, а не по фильму. Есть ли среди пользователей
такие, кто настолько любит кино, что ставит почти исключительно высшие баллы?

```python
%timeit -n 1 -r 1 users_ratings.top_by_ratings(10, metric="average")
users_ratings.top_by_ratings(10, metric="average")
```

Несколько пользователей умудрились поставить средний балл ровно 5.0 -
вероятно, они оценили совсем немного фильмов, и все они оказались любимыми.
А есть ли среди пользователей такие, чьи оценки сильнее всего скачут от
фильма к фильму?

```python
%timeit -n 1 -r 1 users_ratings.top_controversial(10)
users_ratings.top_controversial(10)
```
